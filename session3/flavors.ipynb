```python
import pandas as pd
```

```python
## Read CSV from GitHub url
url = 'https://raw.githubusercontent.com/AlexTheAnalyst/PandasYouTubeSeries/main/Flavors.csv'
df = pd.read_csv(url)
df
```

```python
## Explore the data
# I first check the size, column names, data types, and missing values
# so I know what kind of aggregation is possible.
print(f'Rows: {df.shape[0]}, Columns: {df.shape[1]}')
print('\nColumn names:')
print(df.columns.tolist())
print('\nData types:')
print(df.dtypes)
print('\nMissing values by column:')
print(df.isna().sum())
df.head()
```

```python

base_flavor_groups = df.groupby('Base Flavor')
list(base_flavor_groups.groups.keys())
```

```python
## Make this a new dataframe (grouped object)
grouped_df = df.groupby('Base Flavor')
# size() counts every row in each group, including rows with null values.
group_sizes = grouped_df.size().to_frame(name='Number of Flavors')
group_sizes

```

### My observation

`Vanilla` has 6 rows, while `Chocolate` has 3. I used `size()` first because it directly answers how many rows belong to each group.


```python
# count() works column by column and does not count missing values.
grouped_df.count()
```

```python
## Let's get the means for each Base Flavor
# Only the rating columns are numeric, so I use numeric_only=True.
grouped_df.mean(numeric_only=True)
```

### My observation

The mean `Total Rating` is higher for the Chocolate base-flavor group than for Vanilla in this small dataset. I also noticed that `mean()` should be applied only to numeric columns.


```python
df.groupby('Base Flavor').mean(numeric_only=True)
```

```python
grouped_df.agg(['count', 'sum', 'min', 'max'])

```

```python
grouped_df.agg({
    'Flavor Rating': ['mean', 'min', 'max'],
    'Texture Rating': ['mean', 'min', 'max'],
    'Total Rating': ['count', 'mean', 'sum']
})

## Success? Commit changes to the repository!
```

```python

df.groupby(['Base Flavor', 'Liked']).mean(numeric_only=True)
```

```python

df.groupby(['Base Flavor', 'Liked']).agg({
    'Flavor': 'count',
    'Flavor Rating': ['mean', 'max'],
    'Texture Rating': ['mean', 'max'],
    'Total Rating': ['mean', 'sum']
})
```

### My observation

Grouping by both `Base Flavor` and `Liked` makes the comparison more detailed. There is no Chocolate/No row in this dataset, so pandas only displays combinations that actually exist.


```python

grouped_df.describe()
```

### What I learned

`groupby()` performs the **split** step, functions such as `mean()` or `agg()` perform **apply**, and pandas joins the group results in the **combine** step. Saving the grouped object is convenient when I want to try several calculations on the same groups.
